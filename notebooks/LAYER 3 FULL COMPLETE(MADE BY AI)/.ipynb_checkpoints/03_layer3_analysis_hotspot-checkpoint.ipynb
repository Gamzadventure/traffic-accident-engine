## STEP 3 — Data Quality Check
### Important: NaN vs Zero — They Mean Different Things
| Value | What It Means |
|-------|--------------|
| `NaN` / blank / Not Applicable | Data was **not available** for this city — we could not find it |
| `0` | Data was available and **zero accidents/deaths** happened |

We NEVER mix these two — NaN stays NaN and is shown as "Data N/A" in charts.

````python
print("=" * 60)
print("DATA QUALITY REPORT")
print("=" * 60)

# Key columns check
key_cols = ['accidents_2022','deaths_2022','injured_2022',
            'road_accidents_cases','road_accidents_died','road_danger_index']

print("\nColumn             | NaN count | Zero count | Meaning")
print("-" * 65)
for col in key_cols:
    n_nan  = df[col].isna().sum()
    n_zero = (df[col] == 0).sum() if df[col].dtype != object else 'N/A'
    print(f"  {col:30s} | {n_nan:9} | {str(n_zero):10} |")

print("\n📋 Cities with PARTIAL data (OpenCity Only — road_danger_index = NaN):")
partial = df[df['data_availability'] == 'OpenCity Only']
print(f"   Count: {len(partial)} cities")
print(f"   These cities have accidents/deaths/injured data from OpenCity")
print(f"   but road-level detail (road_accidents_cases etc.) was not available.")
print(f"   We treat this as NaN — NOT as zero.")

print("\n✅ Data integrity confirmed. NaN and Zero are handled separately.")
````

## STEP 4 — Calculate City Risk Score (0 to 100)
> A single number from 0 to 100 that represents how dangerous each city is.  
> Cities with NaN road_danger_index still get a score — we use the available data only.

````python
def normalize_0_100(series):
    mn, mx = series.min(), series.max()
    if mx == mn: return pd.Series([50.0]*len(series), index=series.index)
    return ((series - mn)/(mx - mn))*100

# Normalize core metrics (all 100 cities have these)
df['norm_deaths']    = normalize_0_100(df['deaths_2022'])
df['norm_accidents'] = normalize_0_100(df['accidents_2022'])
df['norm_injured']   = normalize_0_100(df['injured_2022'])

# Road danger index only for cities that have it (52 cities)
df['norm_rdi'] = np.where(
    df['road_danger_index'].notna(),
    normalize_0_100(df['road_danger_index'].fillna(df['road_danger_index'].median())),
    50.0   # neutral score for cities where this data is NaN
)

# Weighted Risk Score
# Deaths 40% + Accidents 30% + Injured 20% + RDI 10%
df['city_risk_score'] = (
    df['norm_deaths']    * 0.40 +
    df['norm_accidents'] * 0.30 +
    df['norm_injured']   * 0.20 +
    df['norm_rdi']       * 0.10
).round(2)

def risk_label(s):
    if s >= 75:   return 'Extreme'
    elif s >= 50: return 'High'
    elif s >= 25: return 'Medium'
    else:         return 'Low'

df['risk_score_label'] = df['city_risk_score'].apply(risk_label)

RISK_COLORS = {
    'Extreme': '#c62828',
    'High':    '#e65100',
    'Medium':  '#f9a825',
    'Low':     '#2e7d32'
}

print("✅ City Risk Scores calculated!")
print("\n📊 Score distribution:")
print(df['risk_score_label'].value_counts().to_string())
print("\n🏆 Top 10 highest risk cities:")
top10 = df[['city','city_risk_score','risk_score_label','deaths_2022','accidents_2022']].nlargest(10,'city_risk_score')
print(top10.to_string(index=False))
````

### Charts 3, 4, 5 — All 100 Cities Risk Score (Split into 3 Professional Charts)
> We sort all 100 cities by risk score and divide into 3 equal groups.  
> **Chart 3** = Cities ranked 1–34 (highest risk)  
> **Chart 4** = Cities ranked 35–67 (medium risk)  
> **Chart 5** = Cities ranked 68–100 (lower risk)  
> This way each chart has ~33 cities — perfectly readable and professional.

````python
df_sorted = df.sort_values('city_risk_score', ascending=False).reset_index(drop=True)
df_sorted['rank'] = df_sorted.index + 1

# Split into 3 groups
group1 = df_sorted.iloc[:34]   # Rank 1-34
group2 = df_sorted.iloc[34:67] # Rank 35-67
group3 = df_sorted.iloc[67:]   # Rank 68-100

def plot_risk_group(data, title, filename, note=""):
    fig, ax = plt.subplots(figsize=(11, len(data)*0.32 + 2))
    colors_bar = [RISK_COLORS[r] for r in data['risk_score_label']]
    bars = ax.barh(data['city'], data['city_risk_score'],
                   color=colors_bar, edgecolor='white', linewidth=0.4)
    for bar, val, rank in zip(bars, data['city_risk_score'], data['rank']):
        ax.text(bar.get_width()+0.4, bar.get_y()+bar.get_height()/2,
                f'#{rank}  {val:.1f}', va='center', fontsize=8, color='#333')
    ax.invert_yaxis()
    ax.set_xlabel('City Risk Score (0–100)', fontsize=11)
    ax.set_xlim(0, 115)
    ax.set_title(title, fontsize=13, fontweight='bold', pad=12)
    if note:
        ax.text(0.99, 0.01, note, transform=ax.transAxes,
                fontsize=8, color='gray', ha='right', va='bottom')
    ax.axvline(75, color='#c62828', linestyle='--', alpha=0.4, linewidth=1)
    ax.axvline(50, color='#e65100', linestyle='--', alpha=0.4, linewidth=1)
    ax.axvline(25, color='#f9a825', linestyle='--', alpha=0.4, linewidth=1)
    legend_els = [mpatches.Patch(color=v,label=k) for k,v in RISK_COLORS.items()]
    ax.legend(handles=legend_els, title='Risk Level', loc='lower right', fontsize=8)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR/filename, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"✅ Saved: {filename}")

plot_risk_group(group1,
    "City Risk Score — Group 1: Highest Risk Cities (Rank 1–34 of 100)",
    "chart03_risk_score_group1_rank1to34.png",
    "Part 1 of 3")

plot_risk_group(group2,
    "City Risk Score — Group 2: Medium Risk Cities (Rank 35–67 of 100)",
    "chart04_risk_score_group2_rank35to67.png",
    "Part 2 of 3")

plot_risk_group(group3,
    "City Risk Score — Group 3: Lower Risk Cities (Rank 68–100 of 100)",
    "chart05_risk_score_group3_rank68to100.png",
    "Part 3 of 3")
````

## STEP 6 — Hotspot Detection (Clustering)
### What is Clustering?
Think of it as **automatic grouping**. You have 100 cities — clustering groups them into  
"families" of similar danger level without you manually deciding the groups.

**K-Means** → Divides into exactly 4 groups (Low / Medium / High / Very High)  
**DBSCAN** → Finds natural clusters automatically + flags **extreme outlier hotspot cities**

````python
features = ['deaths_2022','accidents_2022','injured_2022','city_risk_score']
df_feat  = df[features].fillna(0)

scaler  = StandardScaler()
X_scaled = scaler.fit_transform(df_feat)

# ── K-Means (4 clusters) ──
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df['kmeans_cluster'] = kmeans.fit_predict(X_scaled)

clust_danger = df.groupby('kmeans_cluster')['deaths_2022'].mean().sort_values()
risk_map = {
    clust_danger.index[0]: 'Low Risk',
    clust_danger.index[1]: 'Medium Risk',
    clust_danger.index[2]: 'High Risk',
    clust_danger.index[3]: 'Very High Risk'
}
df['kmeans_label'] = df['kmeans_cluster'].map(risk_map)

# ── DBSCAN (auto detect outliers = extreme hotspots) ──
dbscan = DBSCAN(eps=0.7, min_samples=2)
df['dbscan_cluster'] = dbscan.fit_predict(X_scaled)
df['dbscan_label']   = df['dbscan_cluster'].apply(
    lambda x: 'EXTREME HOTSPOT' if x==-1 else f'Cluster {x+1}')

n_clusters  = len(set(df['dbscan_cluster'])) - (1 if -1 in df['dbscan_cluster'].values else 0)
n_outliers  = (df['dbscan_cluster']==-1).sum()

print("✅ K-Means clustering done!")
print("\n📊 K-Means — Cities per cluster:")
print(df.groupby('kmeans_label')[['deaths_2022','accidents_2022']].mean().round(0).to_string())

print(f"\n✅ DBSCAN clustering done!")
print(f"   Natural clusters found : {n_clusters}")
print(f"   Extreme hotspot cities : {n_outliers}")

print(f"\n🚨 EXTREME HOTSPOT CITIES (most dangerous — stand alone):")
hotspots = df[df['dbscan_cluster']==-1][
    ['city','city_risk_score','deaths_2022','accidents_2022','data_availability']
].sort_values('city_risk_score', ascending=False)
print(hotspots.to_string(index=False))
````

````python
def get_color(score):
    if score>=75:   return '#c62828'
    elif score>=50: return '#e65100'
    elif score>=25: return '#f9a825'
    else:           return '#2e7d32'

# ── Map 1: City circles only ──
city_map = folium.Map(
    location=[20.5937, 78.9629],
    zoom_start=5,
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}',
    attr='Esri'
)

for _,row in df.iterrows():
    city = row['city']
    if city not in CITY_COORDS: continue
    lat,lon   = CITY_COORDS[city]
    score     = row['city_risk_score']
    color     = get_color(score)
    score_s   = f"{score:.1f}"
    deaths_s  = f"{int(row['deaths_2022']):,}"
    acc_s     = f"{int(row['accidents_2022']):,}"
    inj_s     = f"{int(row['injured_2022']):,}"
    avail_s   = str(row['data_availability'])
    risk_s    = str(row['risk_score_label'])

    popup_html = (
        f"<div style='font-family:Arial;min-width:220px'>"
        f"<h4 style='margin:0;color:{color}'>{city}</h4>"
        f"<hr style='margin:4px 0'>"
        f"<b>Risk Score:</b> {score_s}/100 ({risk_s})<br>"
        f"<b>Deaths 2022:</b> {deaths_s}<br>"
        f"<b>Accidents 2022:</b> {acc_s}<br>"
        f"<b>Injured 2022:</b> {inj_s}<br>"
        f"<hr style='margin:4px 0'>"
        f"<b>Data Source:</b> {avail_s}"
        f"</div>"
    )

    folium.CircleMarker(
        location=[lat,lon], radius=6+score/11,
        color='white', weight=1.5,
        fill=True, fill_color=color, fill_opacity=0.88,
        popup=folium.Popup(popup_html, max_width=270),
        tooltip=f"{city} | Risk: {score_s}/100"
    ).add_to(city_map)

legend_html = (
    "<div style='position:fixed;bottom:30px;left:30px;z-index:1000;"
    "background:white;padding:12px;border-radius:8px;"
    "border:2px solid #ccc;font-family:Arial;font-size:13px'>"
    "<b>City Risk Score</b><br>"
    "<span style='color:#c62828'>&#9679;</span> Extreme (75-100)<br>"
    "<span style='color:#e65100'>&#9679;</span> High (50-74)<br>"
    "<span style='color:#f9a825'>&#9679;</span> Medium (25-49)<br>"
    "<span style='color:#2e7d32'>&#9679;</span> Low (0-24)<br>"
    "<small>Circle size = risk level<br>Click city for details</small>"
    "</div>"
)
city_map.get_root().html.add_child(folium.Element(legend_html))
city_map.save(str(OUTPUT_DIR/'map1_city_hotspots.html'))
print("✅ Saved: map1_city_hotspots.html (100 cities — click any city!)")
````

````python
# ── Map 2: Choropleth + City circles combined ──
import json

gadm_to_ours_local = {
    'Andhra Pradesh':'Andhra Pradesh','Bihar':'Bihar',
    'Chhattisgarh':'Chhattisgarh','Chandigarh':'Chandigarh',
    'Gujarat':'Gujarat','Haryana':'Haryana',
    'Jammu and Kashmir':'Jammu & Kashmir','Jharkhand':'Jharkhand',
    'Karnataka':'Karnataka','Kerala':'Kerala',
    'Madhya Pradesh':'Madhya Pradesh','Maharashtra':'Maharashtra',
    'NCT of Delhi':'Delhi','Odisha':'Odisha','Punjab':'Punjab',
    'Rajasthan':'Rajasthan','Tamil Nadu':'Tamil Nadu',
    'Telangana':'Telangana','Uttar Pradesh':'Uttar Pradesh',
    'Uttarakhand':'Uttarakhand','West Bengal':'West Bengal',
    'Assam':'Assam',
}

deaths_lookup = dict(zip(state_summary['state'], state_summary['total_deaths']))

india_states['deaths_val'] = india_states['NAME_1'].map(
    lambda x: deaths_lookup.get(gadm_to_ours_local.get(x, ''), 0)
)

choropleth_map = folium.Map(
    location=[20.5937, 78.9629],
    zoom_start=5,
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}',
    attr='Esri'
)

folium.Choropleth(
    geo_data         = india_states.__geo_interface__,
    data             = india_states[['NAME_1', 'deaths_val']],
    columns          = ['NAME_1', 'deaths_val'],
    key_on           = 'feature.properties.NAME_1',
    fill_color       = 'YlOrRd',
    fill_opacity     = 0.65,
    line_opacity     = 0.5,
    legend_name      = 'Total Road Deaths (2022)',
    nan_fill_color   = '#d9d9d9',
    nan_fill_opacity = 0.4,
).add_to(choropleth_map)

for _, row in df.iterrows():
    city = row['city']
    if city not in CITY_COORDS:
        continue
    lat, lon  = CITY_COORDS[city]
    score     = row['city_risk_score']
    color     = get_color(score)
    score_s   = f"{score:.1f}"
    deaths_s  = f"{int(row['deaths_2022']):,}"
    acc_s     = f"{int(row['accidents_2022']):,}"
    popup_c   = (
        f"<b>{city}</b><br>Risk: {score_s}/100<br>"
        f"Deaths: {deaths_s}<br>Accidents: {acc_s}"
    )
    folium.CircleMarker(
        location     = [lat, lon],
        radius       = 5 + score / 12,
        color        = 'white',
        weight       = 1.2,
        fill         = True,
        fill_color   = color,
        fill_opacity = 0.9,
        popup        = folium.Popup(popup_c, max_width=200),
        tooltip      = f"{city}: {score_s}/100"
    ).add_to(choropleth_map)

choropleth_map.save(str(OUTPUT_DIR / 'map2_choropleth_combined.html'))
print("✅ Saved: map2_choropleth_combined.html")
print("   States colored by deaths + city circles on top!")
````

## STEP 10 — Save All Outputs for Layer 4

````python
# ── Final master dataset ──
final_path = OUTPUT_DIR / 'Layer3_FINAL_100Cities_Master.xlsx'
df.to_excel(final_path, index=False)
print(f"✅ Final dataset saved: {final_path}")

# ── Hotspot cities ──
hotspot_df = df[df['dbscan_cluster']==-1][
    ['city','city_risk_score','risk_score_label','kmeans_label',
     'deaths_2022','accidents_2022','injured_2022',
     'road_danger_index','data_availability','state']
].sort_values('city_risk_score', ascending=False)
hotspot_df.to_excel(OUTPUT_DIR/'Layer3_Hotspot_Cities.xlsx', index=False)
print(f"✅ Hotspot cities saved: {len(hotspot_df)} extreme hotspot cities")

# ── Cluster summary ──
cluster_summary = df.groupby('kmeans_label').agg(
    num_cities      = ('city','count'),
    avg_risk_score  = ('city_risk_score','mean'),
    avg_deaths      = ('deaths_2022','mean'),
    avg_accidents   = ('accidents_2022','mean'),
    total_deaths    = ('deaths_2022','sum'),
    total_accidents = ('accidents_2022','sum'),
).round(1).reset_index()
cluster_summary.to_excel(OUTPUT_DIR/'Layer3_Cluster_Summary.xlsx', index=False)

# ── City coordinates ──
coords_df = pd.DataFrame([
    {'city':c,'latitude':lat,'longitude':lon}
    for c,(lat,lon) in CITY_COORDS.items()
])
coords_df.to_excel(OUTPUT_DIR/'Layer3_City_Coordinates.xlsx', index=False)

print("\n" + "="*65)
print("   LAYER 3 — 100% COMPLETE")
print("="*65)
print()

all_files = sorted(OUTPUT_DIR.iterdir())
print(f"TOTAL OUTPUT FILES: {len(all_files)}")
print()
for f in all_files:
    kb = f.stat().st_size // 1024
    tag = ""
    if 'weather' in f.name:     tag = "  ← OpenWeatherMap API"
    elif 'choropleth' in f.name: tag = "  ← GADM Shapefile"
    elif 'map1' in f.name:       tag = "  ← Main interactive map"
    elif 'FINAL' in f.name:      tag = "  ← INPUT for Layer 4"
    elif 'Hotspot' in f.name:    tag = "  ← Hotspot city list"
    print(f"  {f.name:52s} {kb:>5} KB{tag}")

print()
print("WHAT TO PUSH TO GITHUB:")
print("  git add .")
print('  git commit -m "Layer 3 complete - 100 cities analysis and hotspot detection"')
print("  git push")
print()
print("WHAT YOUR LAYER 4 TEAMMATE NEEDS:")
print("  Layer3_FINAL_100Cities_Master.xlsx")
print("="*65)
````
